```python
import sqlite3
import pandas as pd

pd.set_option('display.max_colwidth', None)

connection = sqlite3.connect("../data/sql-murder-mystery.db")
cursor = connection.cursor()

def sql_query(query):

    cursor.execute(query)
    ans = cursor.fetchall()
    names = [description[0] for description in cursor.description]

    return pd.DataFrame(ans, columns=names)
```

***QUERY 1 — Encontrar el informe del crimen***

```python
query_1 = """
SELECT *
FROM crime_scene_report
WHERE date = 20180115
  AND type = 'murder'
  AND city = 'SQL City';
"""

sql_query(query_1)
```

***QUERY 2 — Encontrar al primer testigo***

```python
query_2 = """
SELECT *
FROM person
WHERE address_street_name = 'Northwestern Dr'
ORDER BY address_number DESC
LIMIT 1;
"""

sql_query(query_2)
```

***QUERY 3 — Buscar la entrevista de Morty***

```python
query_3 = """
SELECT *
FROM interview
WHERE person_id = 14887;
"""

sql_query(query_3)
```

***QUERY 4 — Buscar socios del gimnasio***

```python
query_4 = """
SELECT *
FROM get_fit_now_member
WHERE id LIKE '48Z%'
  AND membership_status = 'gold';
"""

sql_query(query_4)
```

***QUERY 5 — Buscar el coche***

```python
query_5 = """
SELECT *
FROM drivers_license
WHERE plate_number LIKE '%H42W%';
"""

sql_query(query_5)
```

***QUERY 6 — Cruzar sospechosos con sus carnets*** 

```python
query_6 = """
SELECT
    person.id,
    person.name,
    person.license_id,
    drivers_license.plate_number,
    drivers_license.car_make,
    drivers_license.car_model
FROM person
JOIN drivers_license
    ON person.license_id = drivers_license.id
WHERE person.id IN (28819, 67318);
"""

sql_query(query_6)
```
