```python
import os
from dotenv import load_dotenv
from google.cloud import bigquery

load_dotenv()

project_id = os.getenv("GCP_PROJECT_ID")
dataset_id = os.getenv("BQ_DATASET_ID")
credentials_path = os.getenv("GOOGLE_APPLICATION_CREDENTIALS")

client = bigquery.Client.from_service_account_json(
    credentials_path,
    project=project_id
)

print("Conexión a BigQuery establecida correctamente.")
```

```python
def sql_query(query):
    query_job = client.query(query)
    results = query_job.result()

    rows = [dict(row) for row in results]

    for row in rows:
        print(row)

    return rows
```

***Query 1 — Ingresos por mes***

```python
query_1 = """
SELECT
    FORMAT_DATE('%Y-%m', o.order_date) AS month,
    ROUND(
        SUM(
            oi.quantity
            * oi.unit_price
            * (1 - oi.discount)
        ),
        2
    ) AS total_revenue
FROM `tc-sql-507706.tc_sql_dataset.orders` AS o
JOIN `tc-sql-507706.tc_sql_dataset.order_items` AS oi
    ON o.order_id = oi.order_id
WHERE o.order_status != 'cancelled'
GROUP BY month
ORDER BY month
"""

sql_query(query_1)
```

***Query 2 — Productos más vendidos***

```python
query_2 = """
SELECT
    p.product_id,
    p.name AS product_name,
    SUM(oi.quantity) AS units_sold
FROM `tc-sql-507706.tc_sql_dataset.products` AS p
JOIN `tc-sql-507706.tc_sql_dataset.order_items` AS oi
    ON p.product_id = oi.product_id
JOIN `tc-sql-507706.tc_sql_dataset.orders` AS o
    ON oi.order_id = o.order_id
WHERE o.order_status != 'cancelled'
GROUP BY
    p.product_id,
    p.name
ORDER BY units_sold DESC
LIMIT 10
"""

sql_query(query_2)
```

***Query 3 — Clientes por país***

```python
query_3 = """
SELECT
    country,
    COUNT(*) AS number_of_customers
FROM `tc-sql-507706.tc_sql_dataset.customers`
GROUP BY country
ORDER BY number_of_customers DESC
"""

sql_query(query_3)
```

***Query 4 — Tiempo medio de entrega***

```python
query_4 = """
SELECT
    ROUND(
        AVG(
            DATE_DIFF(delivered_date, order_date, DAY)
        ),
        2
    ) AS average_delivery_days
FROM `tc-sql-507706.tc_sql_dataset.orders`
WHERE delivered_date IS NOT NULL
"""

sql_query(query_4)
```

***Query 5 — Ingresos por categoría***

```python
query_5 = """
SELECT
    c.name AS category,
    ROUND(
        SUM(
            oi.quantity
            * oi.unit_price
            * (1 - oi.discount)
        ),
        2
    ) AS total_revenue
FROM `tc-sql-507706.tc_sql_dataset.categories` AS c
JOIN `tc-sql-507706.tc_sql_dataset.products` AS p
    ON c.category_id = p.category_id
JOIN `tc-sql-507706.tc_sql_dataset.order_items` AS oi
    ON p.product_id = oi.product_id
JOIN `tc-sql-507706.tc_sql_dataset.orders` AS o
    ON oi.order_id = o.order_id
WHERE o.order_status != 'cancelled'
GROUP BY category
ORDER BY total_revenue DESC
"""

sql_query(query_5)
```
